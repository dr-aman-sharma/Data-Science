# Web Scraping Project

This is a web scraping project. The idea of the project is to get data from the web with the help of a URL. The data is that of the worlds largest economies in the decreasing order of their GDP. This data is available on a webpage accessible with the following URL: https://web.archive.org/web/20230902185326/https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29

```python
import numpy as np
import pandas as pd
```

```python
URL="https://web.archive.org/web/20230902185326/https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29"
```

```python
tables = pd.read_html(URL)
```

```python
df = tables[3] #Retaining table number 3 as the required dataframe.
df.columns = range(df.shape[1])  # Replacing the column headers with column numbers
```

```python
df.head()
```

```python
df = df[[0,2]]  # Retaining columns with index 0 and 2 (name of country and value of GDP quoted by IMF
df = df.iloc[1:11,:]  # Retaining the Rows with index 1 to 10, indicating the top 10 economies of the world.
df.head()
```

```python
df.columns = ['Country','GDP (Million USD)']   # Assigning column names as "Country" and "GDP (Million USD)"
df['GDP (Million USD)'] = df['GDP (Million USD)'].astype(int)   # Changing the data type of the 'GDP (Million USD)' column to integer. Using astype() method.
df[['GDP (Million USD)']] = df[['GDP (Million USD)']]/1000      # Converting the GDP value in Million USD to Billion USD
df[['GDP (Million USD)']] = np.round(df[['GDP (Million USD)']], 2)    # Using numpy.round() method to round the value to 2 decimal places.
df.rename(columns = {'GDP (Million USD)' : 'GDP (Billion USD)'})     # Renaming the column header from 'GDP (Million USD)' to 'GDP (Billion USD)'

```

```python
df.to_csv('./Largest_economies.csv')    # Loading the DataFrame to the CSV file named "Largest_economies.csv"
```

## Webscraping the stock price
Next we scrape stock data from the web and analyze it. For this we use yfinance library.

```python
import yfinance as yf
import pandas as pd
```

## APPLE

```python
apple = yf.Ticker("AAPL")  # Allows us to extract all the information related to apple stock

apple_share_price_data = apple.history(period="max")   #getting the share proce of the stock
```

```python
apple_share_price_data.head()
```

```python
apple_share_price_data.reset_index(inplace=True)
```

```python
apple_share_price_data.head()
```

```python
import matplotlib.pyplot as plt
apple_share_price_data.plot(x="Date", y="Open")
plt.ylabel("Open  price")
plt.xlabel("Date")
```

```python
apple.dividends.plot()
plt.ylabel("Dividends")
plt.xlabel("Date")
```

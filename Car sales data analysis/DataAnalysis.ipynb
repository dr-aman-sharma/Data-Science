# Car sales price data analysis

We have car sales data available from "https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data". Car's seling price may depend on a number of factors for example engine type, length of the car, fuel system, peak rpm, horse power etc. The aim of this project is to demonstrate cleaning of the data, exploratory data analysis, and development of a model to predict car price. More information about the data is available at "https://archive.ics.uci.edu/dataset/10/automobile".

## Data loading and cleaning

This is the first step in any data science project. I have already downloaded the data file, so I will proceed with loading and cleaning the data.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("auto.csv")

# create headers list
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
print("headers\n", headers)

df.columns = headers
```

```python
print("The first 5 rows of the dataframe") 
df.head(5)
```

```python
df=df.replace('?',np.NaN) # replace the "?" symbol with NaN so the dropna() can remove the missing values
#df=df.dropna(subset=["price"], axis=0)   # drop missing values along the column "price"
```

```python
# describe all the columns in "df" 
df.describe(include = "all")
```

```python
df.info() # It provides a concise summary of the data frame. 
```

```python
missing_data = df.isnull()
missing_data.head(5)
```

```python
# Using a for loop in Python, you can quickly figure out the number of missing values in each column.
for column in missing_data.columns.values.tolist():        
    print(column)
    print (missing_data[column].value_counts())
    print("") 
```

We will now replace the missing data in various columns with the mean

```python
avg_norm_loss = df["normalized-losses"].astype("float").mean(axis=0)
avg_bore=df['bore'].astype('float').mean(axis=0)
avg_horsepower = df['horsepower'].astype('float').mean(axis=0)
avg_peakrpm=df['peak-rpm'].astype('float').mean(axis=0)


df["normalized-losses"] = df["normalized-losses"].fillna(avg_norm_loss)
df['bore'] = df['bore'].fillna(avg_norm_loss)
df['horsepower'] = df['horsepower'].fillna(avg_norm_loss)
df['peak-rpm'] = df['peak-rpm'].fillna(avg_norm_loss)
```

```python
df['num-of-doors'].value_counts()
```

You can see that four doors is the most common type. 

```python
df['num-of-doors'] = df['num-of-doors'].fillna("four")
```

```python
df.dropna(subset=["price"], axis=0, inplace=True)   # simply drop whole row with NaN in "price" column
df.reset_index(drop=True, inplace=True)   # reset index, because we droped two rows
```

```python
df.dtypes   # list of the data types for each column
```

```python
# Converting data types to proper format

df[["bore", "stroke"]] = df[["bore", "stroke"]].astype("float")
df[["normalized-losses"]] = df[["normalized-losses"]].astype("int")
df[["price"]] = df[["price"]].astype("float")
df[["peak-rpm"]] = df[["peak-rpm"]].astype("float")
```

```python
# Convert mpg to L/100km by mathematical operation (235 divided by mpg)
df['city-L/100km'] = 235/df["city-mpg"]

# transform mpg to L/100km by mathematical operation (235 divided by mpg)
df["highway-mpg"] = 235/df["highway-mpg"]

# rename column name from "highway-mpg" to "highway-L/100km"
df.rename(columns={'highway-mpg':'highway-L/100km'}, inplace=True)
```

We finally obtained the cleansed data set with no missing values and with all data in its proper format.

#### Now we normalize length, width and height columns

```python
# replace (original value) by (original value)/(maximum value)
df['length'] = df['length']/df['length'].max()
df['width'] = df['width']/df['width'].max()
df['height'] = df['height']/df['height'].max()
```

```python
len(df["horsepower"].unique())
```

The column "fuel-type" has two unique values: "gas" or "diesel". To use this attribute in regression analysis, we can convert "fuel-type" to indicator variables.

```python
dummy_variable_1 = pd.get_dummies(df["fuel-type"])
dummy_variable_1.rename(columns={'gas':'fuel-type-gas', 'diesel':'fuel-type-diesel'}, inplace=True)
dummy_variable_1.head()
```

```python
# merge data frame "df" and "dummy_variable_1" 
df = pd.concat([df, dummy_variable_1], axis=1)
# drop original column "fuel-type" from "df"
df.drop("fuel-type", axis = 1, inplace=True)
```

```python
# get indicator variables of aspiration and assign it to data frame "dummy_variable_2"
dummy_variable_2 = pd.get_dummies(df['aspiration'])

# change column names for clarity
dummy_variable_2.rename(columns={'std':'aspiration-std', 'turbo': 'aspiration-turbo'}, inplace=True)

# show first 5 instances of data frame "dummy_variable_1"
dummy_variable_2.head()
```

```python
# merge the new dataframe to the original datafram
df = pd.concat([df, dummy_variable_2], axis=1)
# drop original column "aspiration" from "df"
df.drop('aspiration', axis = 1, inplace=True)
```

## Model Development for price prediction

We will now use Model Development to help us predict future observations from the data we have.

#### Model 1: Simple Linear Regression

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
lm
```

```python
X = df[['highway-L/100km']]
Y = df['price']
#highway_mpg_fit
lm.fit(X, Y)
# Find the R^2
print('The R-square is: ', lm.score(X, Y))
```

We can say that ~80.896 % of the variation of price is explained by this simple linear regression.

```python
from sklearn.metrics import mean_squared_error

Yhat=lm.predict(X)
print('The output of the first four predicted value is: ', Yhat[0:4])

mse = mean_squared_error(df['price'], Yhat)
print('The mean square error of price and predicted value is: ', mse)
```

#### Model 2: Multiple Linear Regression

```python
Z = df[['horsepower', 'curb-weight', 'engine-size', 'highway-L/100km']]
Z = Z.astype(float)
lm.fit(Z, df['price'])
print('The R-square is: ', lm.score(Z, df['price']))
```

We can say that ~81.102 % of the variation of price is explained by this multiple linear regression "multi_fit".

```python
Y_predict_multifit = lm.predict(Z)
print('The mean square error of price and predicted value using multifit is: ', \
      mean_squared_error(df['price'], Y_predict_multifit))
```

#### Model 3: Polynomial Fit Regression

```python
from sklearn.metrics import r2_score
```

```python
x = df['highway-L/100km']
y = df['price']
```

```python
# Here we use a polynomial of the 3rd order (cubic) 
f = np.polyfit(x, y, 3)
p = np.poly1d(f)
print(p)
r_squared = r2_score(y, p(x))
print('The R-square value is: ', r_squared)
```

```python
mean_squared_error(df['price'], p(x))
```

We can say that ~67.419 % of the variation of price is explained by this polynomial fit.


If we take a look at the values for the different models we observe the following:

This R-squared in combination with the MSE show that MLR seems like the better model fit in this case compared to SLR.
Usually, more variables lead to better estimate. However, one should always check, MSE and R-squared values to be sure.

#### Simple Linear Model (SLR) vs. Polynomial Fit

MSE: We can see that Polynomial Fit brought down the MSE, since this MSE is smaller than the one from the SLR.

R-squared: The R-squared for the Polynomial Fit is larger than the R-squared for the SLR, so the Polynomial Fit also brought up the R-squared quite a bit.

Since the Polynomial Fit resulted in a lower MSE and a higher R-squared, we can conclude that this was a better fit model than the simple linear regression for predicting "price" with "highway-mpg" as a predictor variable.</p>
